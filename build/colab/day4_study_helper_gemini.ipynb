## ④ 챗봇 실행
아래에 화면이 뜹니다. 멈추려면 셀의 ■(중지)를 누르세요. 코드를 고쳤다면 이 셀만 다시 실행하면 됩니다.

**테스트 순서(1단계 → 3단계)**
1. 「예제 자료로 연습하기」 → 왼쪽 표에 자료 3개가 보이는지
2. 「광합성은 어디서 일어나?」 → 답 아래 **근거**에 파일 · 위치 · 원문이 붙는지
3. 「블랙홀의 온도는?」 → 「자료에서 찾지 못했어요」라고 하는지(지어내지 않는지)
4. 학년을 「초등 3~4학년」으로 바꾸고 같은 질문 → 설명이 쉬워지는지
5. 📝 문제 만들기 → 답을 일부러 틀리게 보내기 → 채점 · 해설 → 🔁 오답 복습
6. ☑ 힌트 모드 → 「광합성과 호흡의 차이는?」 → 정답 대신 힌트가 오는지 → 💡 정답 보기

````python
# ============================================================
# AI 학습 도우미 챗봇 (Gradio + Google Gemini + RAG)
#  - 교과서 · 학습자료(PDF · Word · PPT · TXT)를 올리면 자료를 근거로 답합니다.
#  - 답마다 근거(파일 이름 · 쪽 · 원문 한 구절)를 붙이고, 자료에 없으면 「자료에 없어요」라고 말합니다.
#  - 학년에 맞춰 설명하고, 요약 · 문제 만들기 · 채점 · 오답 복습 · 힌트 모드를 씁니다.
#  - 자료 전체가 아니라 질문과 관련된 조각만 Gemini 로 보냅니다(검색은 코랩 안에서 합니다).
# ============================================================
import os, re, io, json, random, inspect, shutil
import gradio as gr
from google import genai
from google.genai import types

CHUNK = 700          # 근거 조각 하나의 길이(글자)
OVERLAP = 120        # 조각끼리 겹치는 길이(문장이 잘려도 앞뒤가 이어지게)
TOP_K = 6            # 질문 하나에 보내는 조각 수
MIN_SCORE = 0.015    # 이보다 비슷한 조각이 하나도 없으면 「자료에 없어요」(그 밖에는 Gemini 가 판단)

GRADES = ["초등 1~2학년", "초등 3~4학년", "초등 5~6학년", "중학생", "고등학생"]
MODES = ["질문하기", "개념 학습", "문제 풀이"]


# ---------- 1. 자료 읽기 (쪽 번호를 붙여 둡니다) ----------
def ocr_page(page):
    """스캔된 쪽(글자가 그림인 쪽)을 코랩 안에서 글자로 바꿉니다(Tesseract). 외부로 보내지 않습니다."""
    import pytesseract
    from PIL import Image
    pix = page.get_pixmap(dpi=200)
    img = Image.open(io.BytesIO(pix.tobytes("png")))
    # preserve_interword_spaces: 한글 글자 사이에 빈칸이 끼는 것(「광 합 성」)을 막습니다
    return pytesseract.image_to_string(img, lang="kor+eng", config="-c preserve_interword_spaces=1")

def read_pdf(path):
    try:
        import pymupdf as fitz       # PyMuPDF (새 이름)
    except ImportError:
        import fitz
    pages, ocr_n, ocr_fail = [], 0, 0
    with fitz.open(path) as doc:
        for i, page in enumerate(doc, 1):
            text = page.get_text().strip()
            if len(text) < 30:       # 글자가 거의 없으면 스캔본으로 보고 OCR
                try:
                    text = ocr_page(page).strip()
                    if text:
                        ocr_n += 1
                except Exception:
                    ocr_fail += 1
            pages.append((f"{i}쪽", text))
    note = []
    if ocr_n:
        note.append(f"스캔 쪽 {ocr_n}개는 글자 인식(OCR)으로 읽음")
    if ocr_fail:
        note.append(f"스캔 쪽 {ocr_fail}개는 OCR 실패(① 설치 셀을 다시 실행해 보세요)")
    return pages, " · ".join(note)

def read_docx(path):
    import docx
    d = docx.Document(path)
    parts = [p.text for p in d.paragraphs if p.text.strip()]
    for t in d.tables:
        for r in t.rows:
            parts.append(" | ".join(c.text.strip() for c in r.cells))
    # 워드는 쪽 번호가 파일에 없어서 문단 번호로 위치를 적습니다
    return [(f"문단 {i+1}~{min(i+15, len(parts))}", "\n".join(parts[i:i+15]))
            for i in range(0, len(parts), 15)], ""

def read_pptx(path):
    from pptx import Presentation
    out = []
    for i, s in enumerate(Presentation(path).slides, 1):
        t = [sh.text_frame.text for sh in s.shapes if sh.has_text_frame and sh.text_frame.text.strip()]
        if s.has_notes_slide and s.notes_slide.notes_text_frame.text.strip():
            t.append("[발표자 메모] " + s.notes_slide.notes_text_frame.text)
        out.append((f"{i}번 슬라이드", "\n".join(t)))
    return out, ""

def read_txt(path):
    for enc in ("utf-8", "cp949"):
        try:
            with open(path, encoding=enc) as f:
                return [("본문", f.read())], ""
        except UnicodeDecodeError:
            continue
    raise ValueError("글자 인코딩을 알 수 없습니다")

READERS = {".pdf": read_pdf, ".docx": read_docx, ".pptx": read_pptx, ".txt": read_txt, ".md": read_txt}

def make_chunks(name, parts):
    chunks = []
    for loc, text in parts:
        text = re.sub(r"[ \t]+", " ", text).strip()
        step = CHUNK - OVERLAP
        for i in range(0, max(len(text), 1), step):
            piece = text[i:i+CHUNK].strip()
            if len(piece) > 20:
                chunks.append({"file": name, "loc": loc, "text": piece})
            if i + CHUNK >= len(text):
                break
    return chunks


# ---------- 2. 자료 관리 (사람마다 따로 · 창을 닫으면 사라짐) ----------
def new_state():
    return {"docs": {}, "notes": {}, "quiz": None, "wrong": []}

def file_table(state):
    if not state["docs"]:
        return "아직 올린 자료가 없습니다."
    rows = ["| 자료 | 조각 | 글자 수 | 메모 |", "|---|---|---|---|"]
    for name, ch in state["docs"].items():
        rows.append(f"| {name} | {len(ch)} | {sum(len(c['text']) for c in ch):,} | {state['notes'].get(name, '')} |")
    return "\n".join(rows)

def add_files(files, state):
    msgs = []
    for f in files or []:
        path = f if isinstance(f, str) else f.name
        name = os.path.basename(path)
        ext = os.path.splitext(name)[1].lower()
        if ext in (".doc", ".ppt", ".hwp"):
            msgs.append(f"❌ {name}: 옛 형식은 읽지 못합니다. {ext}x(또는 PDF)로 저장해 올려 주세요.")
        elif ext not in READERS:
            msgs.append(f"❌ {name}: 지원하지 않는 형식입니다(PDF · Word(docx) · PPT(pptx) · TXT).")
        else:
            try:
                parts, note = READERS[ext](path)
                ch = make_chunks(name, parts)
                if not ch:
                    msgs.append(f"⚠️ {name}: 읽을 글자가 없습니다.")
                else:
                    state["docs"][name] = ch
                    state["notes"][name] = note
                    msgs.append(f"✅ {name}: 조각 {len(ch)}개" + (f" ({note})" if note else ""))
            except Exception as e:
                msgs.append(f"❌ {name}: 읽기 오류 — {e}")
        try:
            os.remove(path)          # 읽은 뒤 올린 파일 자체는 바로 지웁니다(글자만 이 창에 남김)
        except Exception:
            pass
    names = list(state["docs"])
    return (state, file_table(state), "\n\n".join(msgs),
            gr.update(choices=names, value=None), gr.update(value=None))

def add_samples(state):
    """② 셀에서 만든 예제 자료(SAMPLE_DIR)를 복사해 올립니다 — 내 파일 없이 연습할 때"""
    import tempfile
    folder = os.environ.get("SAMPLE_DIR", "")
    if not folder or not os.path.isdir(folder):
        raise gr.Error("예제 자료가 없습니다. 노트북의 ② 예제 자료 셀을 먼저 실행해 주세요.")
    tmp = tempfile.mkdtemp()
    paths = [shutil.copy(os.path.join(folder, n), os.path.join(tmp, n)) for n in sorted(os.listdir(folder))]
    return add_files(paths, state)[:4]

def remove_file(name, state):
    if name and name in state["docs"]:
        state["docs"].pop(name)
        state["notes"].pop(name, None)
        msg = f"🗑️ {name} 을(를) 목록에서 지웠습니다."
    else:
        msg = "지울 자료를 먼저 고르세요."
    return state, file_table(state), msg, gr.update(choices=list(state["docs"]), value=None)

def clear_all(state):
    return new_state(), "아직 올린 자료가 없습니다.", "모든 자료와 대화를 지웠습니다.", gr.update(choices=[], value=None), []


# ---------- 3. 검색 (RAG): 코랩 안에서 질문과 가까운 조각을 찾습니다 ----------
def all_chunks(state):
    return [c for ch in state["docs"].values() for c in ch]

def search(state, query, k=TOP_K):
    from sklearn.feature_extraction.text import TfidfVectorizer
    from sklearn.metrics.pairwise import cosine_similarity
    ch = all_chunks(state)
    if not ch or not query.strip():
        return [], 0.0
    # 글자 2~3개 단위로 비교 → 한국어 조사(은 · 는 · 이 · 가)가 붙어도 같은 낱말로 잡힙니다
    vec = TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 3), sublinear_tf=True)
    m = vec.fit_transform([c["text"] for c in ch] + [query])
    score = cosine_similarity(m[-1], m[:-1])[0]
    best = sorted(range(len(ch)), key=lambda i: -score[i])[:k]
    return [ch[i] for i in best if score[i] > 0], float(score[best[0]]) if best else 0.0

def spread(state, n=20):
    """요약할 때: 자료마다 처음부터 끝까지 고르게 조각을 뽑습니다"""
    out = []
    per = max(1, n // max(1, len(state["docs"])))
    for ch in state["docs"].values():
        step = max(1, len(ch) // per)
        out += ch[::step][:per]
    return out

def ctx_text(chunks):
    return "\n\n".join(f"<<{c['file']} · {c['loc']}>>\n{c['text']}" for c in chunks)


# ---------- 4. 프롬프트 ----------
BASE_RULES = """너는 학생의 공부를 돕는 친절한 AI 학습 도우미야.
[지켜야 할 것]
1. 아래 [자료] 안의 내용으로만 답해. 자료에 없는 내용은 지어내지 말고 found 를 false 로 하고 「올린 자료에서 찾지 못했어요」라고 말해.
2. 설명은 {grade} 학생이 이해할 수 있는 낱말과 예시로 해. 어려운 낱말은 쉬운 말로 풀어 줘.
3. 답의 근거가 된 곳을 sources 에 적어. file 과 loc 는 [자료]의 <<파일 · 위치>> 를 그대로 쓰고, quote 는 자료의 원문을 한 글자도 바꾸지 말고 20~60자 그대로 옮겨.
4. 정답을 외우게 하기보다 왜 그런지 이해하도록 설명해.
5. 한국어로, 너무 길지 않게 답해. 마크다운(굵게 · 번호 목록)을 써도 좋아.
반드시 아래 JSON 하나로만 답해:
{{"answer": "학생에게 보여 줄 답(마크다운)", "found": true, "sources": [{{"file": "", "loc": "", "quote": ""}}],
 "answer_key": [], "graded": false, "wrong": []}}"""

TASKS = {
    "질문하기": "학생의 질문에 자료를 근거로 답해. 답 끝에 학생이 이해했는지 확인하는 짧은 질문을 하나 덧붙여.",
    "개념 학습": ("학생이 물은 개념을 ① 한 줄 뜻 ② 생활 속 쉬운 예시나 비유 ③ 자료에 나온 핵심 내용 순서로 설명해. "
               "끝에 이해를 확인하는 질문을 하나 해."),
    "문제 풀이": ("학생이 문제를 풀다가 묻는 거야. 풀이 방법을 단계로 나눠 설명해."),
    "요약": "올린 자료 전체를 시험 대비용으로 정리해. 단원(또는 자료)별로 핵심 개념 3~5개를 짧은 문장으로, 꼭 외울 낱말은 **굵게** 표시해.",
    "문제": ("자료 내용으로 문제 3개를 만들어(객관식 1개 · OX 1개 · 단답형 1개). answer 에는 문제만 보여 주고 정답은 쓰지 마. "
           "문제 끝에 「답을 적어 보내면 채점해 줄게요」라고 써. "
           "answer_key 에 [{\"no\": 1, \"type\": \"객관식\", \"question\": \"\", \"answer\": \"\", \"explain\": \"\", \"concept\": \"\", \"loc\": \"파일 · 위치\"}] 형식으로 정답을 적어."),
    "채점": ("학생이 앞에서 낸 문제에 답을 보냈어. [정답표]와 비교해 문제마다 맞음/틀림을 알려 주고 해설을 해 줘. graded 를 true 로 해. "
           "틀린 문제는 wrong 에 [{\"concept\": \"틀린 개념\", \"question\": \"문제\"}] 로 적어. "
           "학생의 글이 답이 아니라 다른 질문이면 graded 를 false 로 하고 그 질문에 답해."),
    "오답": ("[틀린 개념] 목록을 보고, 개념마다 왜 헷갈리기 쉬운지와 바른 뜻을 다시 쉽게 설명해. "
           "그다음 비슷한 새 문제를 개념마다 1개씩 내고 정답은 쓰지 마. answer_key 는 「문제」 때와 같은 형식으로 적어."),
    "정답": "바로 앞에서 힌트만 주었던 질문의 최종 정답과 해설을 알려 줘.",
}

HINT_RULE = ("\n[힌트 모드] 정답이나 결론을 바로 말하지 마. 학생이 스스로 생각하도록 힌트 1개와 생각을 이끄는 질문 1~2개만 줘. "
             "채점할 때도 틀린 문제의 정답은 말하지 말고 힌트를 주고 다시 풀어 보게 해. "
             "학생이 「정답 보기」를 누르면 그때 알려 줘.")


# ---------- 5. Gemini 부르기 ----------
def get_client(key_box):
    key = (key_box or "").strip() or os.environ.get("GOOGLE_API_KEY", "") or os.environ.get("GEMINI_API_KEY", "")
    if not key:
        raise gr.Error("Gemini API 키가 없습니다. 왼쪽 열쇠(보안 비밀)에 GOOGLE_API_KEY 로 넣거나, 위 「API 키」 칸에 붙여 넣어 주세요.")
    return genai.Client(api_key=key)

def friendly(e):
    s = str(e)
    if "API_KEY_INVALID" in s or "API key not valid" in s:
        return "API 키가 맞지 않습니다. Google AI Studio 에서 받은 키를 다시 확인해 주세요."
    if "RESOURCE_EXHAUSTED" in s or "429" in s:
        return "무료 사용량을 잠시 넘었습니다. 1분쯤 기다렸다가 다시 눌러 주세요."
    if "UNAVAILABLE" in s or "503" in s or "overloaded" in s:
        return "Gemini 서버가 바쁩니다. 잠시 뒤 다시 눌러 주세요."
    if "PERMISSION_DENIED" in s or "403" in s:
        return "이 키로는 이 모델을 쓸 수 없습니다. 위 「모델」에서 다른 모델을 골라 보세요."
    return "오류가 났습니다: " + s[:300]

def pick_model(client, wanted):
    """적은 모델이 없으면(이름이 바뀌었으면) 쓸 수 있는 flash 모델을 찾아 씁니다"""
    try:
        names = [m.name.replace("models/", "") for m in client.models.list()
                 if "generateContent" in (getattr(m, "supported_actions", None) or ["generateContent"])]
    except Exception:
        return wanted
    if wanted in names:
        return wanted
    flash = sorted([n for n in names if "flash" in n and "image" not in n and "tts" not in n
                    and "live" not in n], reverse=True)
    return flash[0] if flash else wanted

def call_gemini(client, model, system, history, user_text):
    contents = []
    for m in history[-6:]:                       # 앞 대화 3번까지 기억(이어지는 질문용)
        role = "user" if m["role"] == "user" else "model"
        contents.append({"role": role, "parts": [{"text": to_text(m["content"])[:3000]}]})
    contents.append({"role": "user", "parts": [{"text": user_text}]})
    cfg = types.GenerateContentConfig(system_instruction=system, temperature=0.3,
                                      response_mime_type="application/json")
    try:
        r = client.models.generate_content(model=model, contents=contents, config=cfg)
    except Exception as e:
        if "NOT_FOUND" in str(e) or "404" in str(e):
            model = pick_model(client, model)
            r = client.models.generate_content(model=model, contents=contents, config=cfg)
        else:
            raise
    raw = (r.text or "").strip()
    raw = re.sub(r"^```(?:json)?|```$", "", raw).strip()
    try:
        return json.loads(raw)
    except json.JSONDecodeError:
        return {"answer": raw or "답을 만들지 못했어요. 다시 한 번 눌러 주세요.", "found": True, "sources": []}


# ---------- 6. 답 보여 주기 (근거가 진짜 자료에 있는지 한 번 더 확인) ----------
def norm(s):
    return re.sub(r"\s+", "", s or "")

def check_quote(state, src):
    q = norm(src.get("quote"))
    if len(q) < 5:
        return False
    return any(q in norm(c["text"]) for c in state["docs"].get(src.get("file"), []))

def render(state, data):
    text = (data.get("answer") or "").strip()
    srcs = [s for s in (data.get("sources") or []) if isinstance(s, dict)]
    if srcs:
        lines = []
        for s in srcs[:4]:
            mark = "" if check_quote(state, s) else " ⚠️ 원문과 글자가 달라요 — 자료에서 직접 확인해 보세요"
            lines.append(f"- 📄 **{s.get('file','')}** · {s.get('loc','')} — 「{s.get('quote','')}」{mark}")
        text += "\n\n**근거**\n" + "\n".join(lines)
    return text

def to_text(content):
    # Gradio 6 은 대화 내용을 [{"type": "text", "text": ...}] 목록으로 돌려줍니다
    if isinstance(content, str):
        return content
    if isinstance(content, list):
        return "\n".join(c.get("text", "") if isinstance(c, dict) else str(c) for c in content)
    return str(content or "")


# ---------- 7. 단추마다 할 일 ----------
def run(action, text, grade, mode, hint, history, state, key_box, model):
    text = (text or "").strip()
    history = history or []
    if not state["docs"]:
        raise gr.Error("먼저 왼쪽에 학습자료를 올려 주세요.")
    if action == "chat" and not text:
        raise gr.Error("질문을 적어 주세요.")

    # (1) 어떤 일을 할지 정하기
    if action == "chat":
        task = "채점" if state["quiz"] else mode
    else:
        task = action
    shown = {"요약": "📌 요약해 줘", "문제": "📝 문제 만들어 줘" + (f" ({text})" if text else ""),
             "오답": "🔁 오답 복습할래", "정답": "💡 정답 보여 줘"}.get(task, text)

    if task == "오답" and not state["wrong"]:
        msg = "아직 틀린 문제가 없어요. 「문제 만들기」로 먼저 풀어 볼까요?"
        return "", history + [{"role": "user", "content": shown}, {"role": "assistant", "content": msg}], state

    # (2) 관련 자료 찾기 — 자료 전체가 아니라 필요한 조각만 보냅니다
    extra = ""
    if task == "요약":
        chunks = spread(state, 20)
    elif task == "문제":
        chunks = search(state, text)[0] if text else random.sample(all_chunks(state), min(TOP_K, len(all_chunks(state))))
    elif task == "오답":
        q = " ".join(w["concept"] + " " + w.get("question", "") for w in state["wrong"][-5:])
        chunks = search(state, q)[0]
        extra = "[틀린 개념]\n" + json.dumps(state["wrong"][-5:], ensure_ascii=False)
    elif task == "채점":
        q = " ".join(a.get("concept", "") + " " + a.get("question", "") for a in state["quiz"])
        chunks = search(state, q + " " + text)[0]
        extra = "[정답표]\n" + json.dumps(state["quiz"], ensure_ascii=False)
    elif task == "정답":
        last_q = next((to_text(m["content"]) for m in reversed(history) if m["role"] == "user"), "")
        chunks = search(state, last_q)[0]
    else:
        # 이어지는 질문(「그건 왜?」)도 찾을 수 있게 바로 앞 질문을 함께 넣어 찾습니다
        prev = next((to_text(m["content"]) for m in reversed(history) if m["role"] == "user"), "")
        chunks, best = search(state, text + " " + prev)
        if best < MIN_SCORE:
            msg = ("올린 자료에서 이 질문과 관련된 내용을 찾지 못했어요. 🙂\n"
                   "- 자료에 나오는 낱말로 바꿔 물어보거나\n- 관련 자료를 더 올려 주세요.")
            return "", history + [{"role": "user", "content": text}, {"role": "assistant", "content": msg}], state

    # (3) Gemini 에게 묻기
    system = BASE_RULES.format(grade=grade) + "\n[할 일] " + TASKS[task] + (HINT_RULE if hint and task != "정답" else "")
    user_text = f"[자료]\n{ctx_text(chunks)}\n\n{extra}\n\n[학생]\n{text or shown}"
    try:
        client = get_client(key_box)
        data = call_gemini(client, model, system, history, user_text)
    except gr.Error:
        raise
    except Exception as e:
        raise gr.Error(friendly(e))

    # (4) 문제 · 채점 기록 남기기
    if task in ("문제", "오답") and data.get("answer_key"):
        state["quiz"] = data["answer_key"]
    if task == "채점" and data.get("graded"):
        state["wrong"] += [w for w in (data.get("wrong") or []) if isinstance(w, dict) and w.get("concept")]
        state["quiz"] = None if not hint or not data.get("wrong") else state["quiz"]   # 힌트 모드는 틀리면 다시 풀기

    answer = render(state, data)
    if data.get("found") is False:
        answer += "\n\n_(올린 자료에서 근거를 찾지 못했어요. 자료에 나오는 낱말로 바꿔 물어보거나 자료를 더 올려 주세요.)_"
    history = history + [{"role": "user", "content": shown}, {"role": "assistant", "content": answer}]
    return "", history, state


# ---------- 8. 화면 ----------
HOWTO = """### 시작하는 방법
1. **Gemini API 키**를 준비합니다 → [Google AI Studio](https://aistudio.google.com/apikey)에서 무료로 받습니다. 코랩 왼쪽 🔑(보안 비밀)에 `GOOGLE_API_KEY` 로 넣었다면 아래 칸은 비워 둡니다.
2. 왼쪽에 **학습자료**(PDF · Word · PPT · TXT)를 올립니다. 여러 개를 한꺼번에 올려도 됩니다.
3. 위에서 **학년**과 **학습 모드**를 고르고, 아래 칸에 질문을 적어 Enter 를 누릅니다.
4. **요약하기 · 문제 만들기 · 오답 복습** 단추도 눌러 보세요. 문제를 받으면 답을 적어 보내면 채점해 줍니다.

예시: 「광합성이 뭐야?」 · 「3쪽 내용을 쉽게 설명해줘」 · (문제를 받은 뒤) 「1번 ③, 2번 O, 3번 엽록체」"""

chat_kwargs = {"label": "대화", "height": 520}
if "type" in inspect.signature(gr.Chatbot.__init__).parameters:   # Gradio 4~5 는 messages 형식을 직접 지정
    chat_kwargs["type"] = "messages"
blocks_kwargs = {"title": "AI 학습 도우미"}
if "delete_cache" in inspect.signature(gr.Blocks.__init__).parameters:
    blocks_kwargs["delete_cache"] = (600, 1800)   # 올린 임시 파일을 30분 지나면 지웁니다

with gr.Blocks(**blocks_kwargs) as demo:
    state = gr.State(new_state())                 # 이 창(사람)만의 자료 · 문제 · 오답 기록
    gr.Markdown("# 📚 AI 학습 도우미")
    with gr.Accordion("시작하는 방법", open=True):
        gr.Markdown(HOWTO)
    with gr.Row():
        grade = gr.Dropdown(GRADES, value="중학생", label="학년", scale=1)
        mode = gr.Radio(MODES, value="질문하기", label="학습 모드", scale=2)
        hint = gr.Checkbox(label="힌트 모드 (정답을 바로 알려 주지 않아요)", value=False, scale=1)
    with gr.Accordion("API 키 · 모델", open=False):
        with gr.Row():
            key_box = gr.Textbox(label="API 키", type="password", placeholder="보안 비밀에 넣었다면 비워 두세요", scale=3)
            model = gr.Dropdown(["gemini-2.5-flash", "gemini-2.5-flash-lite", "gemini-2.5-pro"],
                                value="gemini-2.5-flash", label="모델", allow_custom_value=True, scale=1)
    with gr.Row():
        with gr.Column(scale=1, min_width=280):
            files = gr.File(label="학습자료 올리기", file_count="multiple", type="filepath",
                            file_types=[".pdf", ".docx", ".pptx", ".txt", ".md"])
            upload_msg = gr.Markdown()
            table = gr.Markdown("아직 올린 자료가 없습니다.")
            with gr.Row():
                pick = gr.Dropdown([], label="지울 자료", scale=2)
                del_btn = gr.Button("삭제", size="sm", scale=1)
            sample_btn = gr.Button("예제 자료로 연습하기", size="sm")
            clear_btn = gr.Button("모두 지우기(자료 · 대화 · 기록)", size="sm", variant="stop")
        with gr.Column(scale=3):
            chat = gr.Chatbot(**chat_kwargs)
            with gr.Row():
                b_sum = gr.Button("📌 요약하기", size="sm")
                b_quiz = gr.Button("📝 문제 만들기", size="sm")
                b_wrong = gr.Button("🔁 오답 복습", size="sm")
                b_ans = gr.Button("💡 정답 보기", size="sm")
            with gr.Row():
                q_box = gr.Textbox(label="질문 (문제 만들기 때는 단원 · 주제를 적으면 그 부분으로 만들어요)",
                                   placeholder="궁금한 것을 적고 Enter", scale=5)
                send = gr.Button("보내기", variant="primary", scale=1)

    common = [q_box, grade, mode, hint, chat, state, key_box, model]
    outs = [q_box, chat, state]
    files.upload(add_files, [files, state], [state, table, upload_msg, pick, files])
    sample_btn.click(add_samples, state, [state, table, upload_msg, pick])
    del_btn.click(remove_file, [pick, state], [state, table, upload_msg, pick])
    clear_btn.click(clear_all, state, [state, table, upload_msg, pick, chat])
    q_box.submit(lambda *a: run("chat", *a), common, outs)
    send.click(lambda *a: run("chat", *a), common, outs)
    b_sum.click(lambda *a: run("요약", *a), common, outs)
    b_quiz.click(lambda *a: run("문제", *a), common, outs)
    b_wrong.click(lambda *a: run("오답", *a), common, outs)
    b_ans.click(lambda *a: run("정답", *a), common, outs)

if __name__ == "__main__":
    demo.launch(share=True, debug=True)
````
